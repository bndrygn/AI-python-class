```python
import matplotlib.pyplot as plt

preci = {'spring': 330.5, 'summer': 612.8, 'autumn': 256.4, 'winter': 13.3}
print(preci.keys())                          # 항목 이름 = 키
print(preci.values())                        # 항목 값 = 값

plt.figure(figsize=(5, 3))
plt.bar(preci.keys(), preci.values())        # keys()와 values()를 그대로 넣는다
plt.title('precipitation')
plt.ylabel('mm')
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [0, 1, 6, 3, 4, 5, 6]
y = [0, 2, 15, 17, 25, 30, 25]

plt.figure(figsize=(9, 2.5))

plt.subplot(1, 3, 1)
plt.scatter(x, y, c='m', marker='^')                         # c: 색, marker: 점 모양
plt.title("c='m', marker='^'")

plt.subplot(1, 3, 2)
plt.scatter(x, y, s=9, c='lightslategray', marker='x')       # s: 점 크기
plt.title("s=9, marker='x'")

plt.subplot(1, 3, 3)
plt.scatter(x, y, c='#009070', marker='*')
plt.title("c='#009070', marker='*'")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

temperature = [6, 2.2, 4.6, 7.7, 8, 9.2, 11.7, 12.5, 10.3, 13.5,
               15.2, 5.9, 2.9, 9.2, 8.6, 4.6, 7.2, 7.7, 8.3, 9.5,
               13, 17, 17.4, 12.5, 12.5, 9.4, 7.3, 9.2, 11.8, 13.3, 15]
print('일 수:', len(temperature))

plt.figure(figsize=(8, 3))

plt.subplot(1, 2, 1)
count, edges, _ = plt.hist(temperature)       # 기본 10개 구간. 구간별 도수와 경계값을 돌려준다
plt.title('default (10 bins)')
print('구간별 도수:', [int(c) for c in count])

plt.subplot(1, 2, 2)
plt.hist(temperature, bins=15, alpha=0.5)    # bins: 구간 수, alpha: 투명도
plt.xlabel('temperature')
plt.ylabel('frequency')
plt.title('bins=15, alpha=0.5')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# [코드 11-32] (교재 3행의 'Ryan',,'Alex' 는 쉼표가 두 개라 오류이므로 하나로 쓴다)
df1 = pd.DataFrame({'name': ['Jessica', 'Liam', 'Sophia', 'Ryan', 'Alex'],
                    'gender': ['F', 'M', 'F', 'M', 'F'],
                    'age': [21, 23, 20, 22, 20]})
df2 = pd.DataFrame({'name': ['Sophia', 'Liam', 'Ryan', 'Alex', 'Jessica'],      # 순서가 df1 과 다르다
                    'department': ['computer science', 'mathematics', 'law',
                                   'computer science', 'law']})

df3 = pd.merge(df1, df2)                     # [코드 11-33] 이름이 같은 열(name)을 기준으로 결합
print(df3)

df3 = pd.merge(df1, df2, on='name')          # [코드 11-34] 기준 열을 on 으로 직접 지정 (권장)
print(df3)

# [코드 11-35] 학과별 학과장 정보
df4 = pd.DataFrame({'department': ['computer science', 'mathematics', 'law'],
                    'head of department': ['Steve', 'Emma', 'Carly']})
df5 = pd.merge(df3, df4, on='department')    # [코드 11-36] department 기준으로 또 결합
print(df5)
```
