```python
#숫자를 입력받아서 짝수인지 흘수인지 출력하는 프로그램을 작성하시오
num=int(input('숫자를 입력하세요'))
num=num%2
if num>=0:
    print(num,'은/는 흘수입니다')
else:
    print(num,'은/는 짝수입니다')
    
```
