```python
email = input("이메일 주소를 입력하세요: ")

if "@" in email:
    id, domain = email.split("@")
    print("아이디:", id)
    print("도메인:", domain)
else:
    print("올바른 이메일이 아닙니다.")
```

```python
email = input("이메일 주소를 입력하세요: ")

at_index = email.find("@")

if at_index != -1:
    id = email[:at_index]
    domain = email[at_index + 1:]
    print("아이디:", id)
    print("도메인:", domain)
else:
    print("올바른 이메일이 아닙니다.")
```

```python
langs = {'Python':0,'C':0,'Java':0,'JavaScript':0}

while True:
    s = input('선호하는 언어(Python,C,Java,JavaScript) 또는 종료: ')
    if s=='종료':
        print('종료합니다.')
        break
    elif s in langs:
        langs[s] += 1
    else:
        print(f'{s}는 등록되지 않은 언어입니다.')

for key, value in langs.items():
    print(key, value)
```

```python
capitals= {
    "India": "New Delhi",
    "France": "Paris",
    "Japan": "Tokyo",
    "South Korea": "Seoul"
}
print("국가의 수도를 맞춰보세요! (종료하려면 'q'를 입력하세요)")

for country, correct_capital in capitals.items():
    print(f"\n[질문] {country}의 수도는 어디입니까?")

    user_input = input("답: ")

    # 종료 조건
    if user_input.strip().lower() == 'q':
        print("프로그램을 종료합니다.")
        break

    # 입력값과 정답을 정리
    user_input_clean = user_input.strip().lower()
    correct_capital_clean = correct_capital.strip().lower()

    # 정답 확인
    if user_input_clean == correct_capital_clean:
        print("정답입니다!")
    else:
        print(f">> 틀렸습니다. 정답은 [{correct_capital}]입니다. <<")

print("\n수도 퀴즈가 끝났습니다!")
```
